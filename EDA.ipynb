```python
from pathlib import Path
import pandas as pd
import pyarrow.parquet as pq

folder = Path("Data/Goodreads_Books/genres_top100")

all_dfs = []

for file in folder.glob("*.parquet"):
    table = pq.read_table(file)
    df = table.to_pandas()
    
    # file.stem gives the filename without the .parquet extension
    df["genre"] = file.stem
    
    all_dfs.append(df)

# Combine all genre dataframes into one
combined_df = pd.concat(all_dfs, ignore_index=True)

# Save to CSV if you want a file output
combined_df.to_csv("Data/Goodreads_Books/all_genres_combined.csv", index=False)

combined_df.head()
```
